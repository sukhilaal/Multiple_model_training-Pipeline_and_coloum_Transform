```python
cat_cause = ['sex', 'smoker', 'day']
num_cause = ['total_bill', 'tip', 'size']
```

```python
# feature enginnering automation using pipeline and coloum transformer
numerical_pipeline = Pipeline(steps =[('imputation',SimpleImputer(strategy = 'median')),('scaling',StandardScaler),()])
```

```python
cat_pipeline = Pipeline(steps =[('imputation',SimpleImputer(strategy = 'most_frequent')),('onehot',OneHotEncoder()),()])
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Re-define pipelines and preprocessor with correct instantiation and without extra tuples
numerical_pipeline = Pipeline(steps=[('imputation', SimpleImputer(strategy='median')), ('scaling', StandardScaler())])
cat_pipeline = Pipeline(steps=[('imputation', SimpleImputer(strategy='most_frequent')), ('onehot', OneHotEncoder(handle_unknown='ignore'))])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_pipeline, num_cause),
        ('cat', cat_pipeline, cat_cause)
    ])

# x_train and x_test are now guaranteed to be DataFrames from the split, so the explicit DataFrame conversion check is no longer needed here.
# It was causing an issue when x_train was already a transformed numpy array from a previous run.

x_train = preprocessor.fit_transform(x_train)
x_test = preprocessor.transform(x_test)
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Initialize the Logistic Regression model
model = LogisticRegression(solver='liblinear', random_state=42)

# Train the model
model.fit(x_train, y_train)

print("Logistic Regression model trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Make predictions on the test set
y_pred = model.predict(x_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.metrics import accuracy_score, classification_report

def model_train_eval(models, x_train, y_train, x_test, y_test):
    results = {}
    for name, model in models.items():
        print(f"\nTraining {name}...")
        # Train the model
        model.fit(x_train, y_train)

        # Make predictions
        y_pred = model.predict(x_test)

        # Evaluate the model
        accuracy = accuracy_score(y_test, y_pred)
        report = classification_report(y_test, y_pred, output_dict=True)

        results[name] = {
            'accuracy': accuracy,
            'classification_report': report
        }

        print(f"{name} Accuracy: {accuracy:.4f}")
        print(f"Classification Report for {name}:\n{classification_report(y_test, y_pred)}")

    print("\nAll models trained and evaluated. Results stored in 'results' dictionary.")
    return results

# Call the function with your defined models and data
model_results = model_train_eval(models, x_train, y_train, x_test, y_test)
print("\nOverall model results:")
for name, res in model_results.items():
    print(f"  {name}: Accuracy = {res['accuracy']:.4f}")
```
